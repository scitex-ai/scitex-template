# Class counts of the training subset (notebook variant)

The same contract as a script in `scripts/`: the work is a `@scitex_session.session`
function, every read/write goes through scitex-io, and parameters come from
`config/*.yaml`. `make notebook` executes it top to bottom with
`clew execute-notebook` (so the run is recorded, with this notebook as its code)
and then checks it with `scitex-notebook verify`.

```python
import numpy as np
import scitex_io as sio
import scitex_session as sess
```

```python
@sess.session
def main(CONFIG=sess.INJECTED, logger=sess.INJECTED):
    """Count training images per digit class."""
    y_train = sio.load(CONFIG.PATH.MNIST.Y_TRAIN)
    digits, counts = np.unique(y_train, return_counts=True)
    class_counts = {str(d): int(c) for d, c in zip(digits, counts)}
    sio.save(class_counts, CONFIG.PATH.MNIST.CLASS_COUNTS, symlink_from_cwd=True)
    logger.info(f"class counts: {class_counts}")
    return 0
```

```python
main()
```
